# Forward Propagation

---

## 1. What is Forward Propagation?

**Forward propagation** is the process of passing an input through a neural network layer by layer — from the input all the way to the final prediction. It is called "forward" because the data flows in one direction: left to right, input → hidden layers → output.

At each node, two things happen:
1. **Weighted sum** — multiply each incoming value by its weight and add a bias
2. **Activation** — pass the result through a non-linear function (sigmoid, ReLU, etc.)

The output of one layer becomes the input to the next. This chain of transformations is what allows the network to model complex, non-linear relationships.

> Forward propagation does **not** update any weights — it only computes a prediction. Weight updates happen during **backpropagation** (the reverse pass).

---

## 2. The Prediction Formula

For any single node, the computation is:

$$o = \sigma(w^T x + b)$$

Where:
- `x` = input vector to that node
- `w` = weight vector for that node
- `b` = bias of that node
- `σ` = sigmoid activation function: $\sigma(z) = \frac{1}{1 + e^{-z}}$

For an entire layer at once, written in matrix form:

$$A^{[l]} = \sigma\left(A^{[l-1]} \cdot W^{[l]} + b^{[l]}\right)$$

Where:
- $A^{[l-1]}$ = output matrix of the previous layer (or the input `X` for layer 1)
- $W^{[l]}$ = weight matrix of layer `l`
- $b^{[l]}$ = bias vector of layer `l`
- $A^{[l]}$ = output of layer `l`, which becomes input to layer `l+1`

---

## 5. Weight Matrix Notation

Weights going into layer 1 are organized as `W¹`, a matrix of shape **(4 × 3)**:

$$W^1 = \begin{bmatrix} w^1_{11} & w^1_{12} & w^1_{13} \\ w^1_{21} & w^1_{22} & w^1_{23} \\ w^1_{31} & w^1_{32} & w^1_{33} \\ w^1_{41} & w^1_{42} & w^1_{43} \end{bmatrix}$$

Each **column** is the weight vector for one hidden node. So column 1 `[w¹₁₁, w¹₂₁, w¹₃₁, w¹₄₁]` connects all 4 inputs to node `b₁₁`.

Similarly:
- `W²` has shape **(3 × 2)** — connects L1 (3 nodes) to L2 (2 nodes)
- `W³` has shape **(2 × 1)** — connects L2 (2 nodes) to L3 (1 node)

Bias vectors:
- `b¹ = [b₁₁, b₁₂, b₁₃]` — shape (1 × 3)
- `b² = [b₂₁, b₂₂]` — shape (1 × 2)
- `b³ = [b₃₁]` — shape (1 × 1)

## 6. Worked Example — Step by Step

We use small placeholder weights to show the mechanics. In practice, these are initialized randomly and learned via backpropagation.

**Assumed weights (for illustration):**

$$W^1 = \begin{bmatrix} 0.1 & 0.2 & 0.3 \\ 0.1 & 0.1 & 0.1 \\ 0.2 & 0.1 & 0.2 \\ 0.1 & 0.2 & 0.1 \end{bmatrix}, \quad b^1 = [0.1,\ 0.1,\ 0.1]$$

$$W^2 = \begin{bmatrix} 0.3 & 0.2 \\ 0.2 & 0.3 \\ 0.1 & 0.1 \end{bmatrix}, \quad b^2 = [0.1,\ 0.1]$$

$$W^3 = \begin{bmatrix} 0.5 \\ 0.5 \end{bmatrix}, \quad b^3 = [0.1]$$

Input row: `A⁰ = [8.1, 92, 75, 76]`

---

### Layer 1 — L0 → L1

Compute `Z¹ = A⁰ · W¹ + b¹`:

**Node b₁₁:**
```
z = 8.1(0.1) + 92(0.1) + 75(0.2) + 76(0.1) + 0.1
  = 0.81 + 9.2 + 15.0 + 7.6 + 0.1
  = 32.71
o₁₁ = σ(32.71) ≈ 1.0000
```

**Node b₁₂:**
```
z = 8.1(0.2) + 92(0.1) + 75(0.1) + 76(0.2) + 0.1
  = 1.62 + 9.2 + 7.5 + 15.2 + 0.1
  = 33.62
o₁₂ = σ(33.62) ≈ 1.0000
```

**Node b₁₃:**
```
z = 8.1(0.3) + 92(0.1) + 75(0.2) + 76(0.1) + 0.1
  = 2.43 + 9.2 + 15.0 + 7.6 + 0.1
  = 34.33
o₁₃ = σ(34.33) ≈ 1.0000
```

$$A^{[1]} = [o_{11},\ o_{12},\ o_{13}] \approx [1.0,\ 1.0,\ 1.0]$$

> The large input values drive z very high, saturating sigmoid to ≈ 1. In real networks, inputs are **normalized** (e.g. scaled to 0–1) before training to prevent this.

---

### Layer 2 — L1 → L2

Input is now `A¹ = [1.0, 1.0, 1.0]`. Compute `Z² = A¹ · W² + b²`:

**Node b₂₁:**
```
z = 1.0(0.3) + 1.0(0.2) + 1.0(0.1) + 0.1
  = 0.3 + 0.2 + 0.1 + 0.1
  = 0.7
o₂₁ = σ(0.7) = 1 / (1 + e^{-0.7}) ≈ 0.6682
```

**Node b₂₂:**
```
z = 1.0(0.2) + 1.0(0.3) + 1.0(0.1) + 0.1
  = 0.2 + 0.3 + 0.1 + 0.1
  = 0.7
o₂₂ = σ(0.7) ≈ 0.6682
```

$$A^{[2]} = [o_{21},\ o_{22}] \approx [0.6682,\ 0.6682]$$

---

### Layer 3 — L2 → L3 (Output)

Input is `A² = [0.6682, 0.6682]`. Compute `Z³ = A² · W³ + b³`:

```
z = 0.6682(0.5) + 0.6682(0.5) + 0.1
  = 0.3341 + 0.3341 + 0.1
  = 0.7682

o₃₁ = σ(0.7682) = 1 / (1 + e^{-0.7682}) ≈ 0.6832
```

$$A^{[3]} = o_{31} = \hat{y} \approx 0.6832$$

---

## 7. Making the Final Prediction

The output `ŷ = 0.6832` is a **probability** — the model thinks there is a ~68% chance this student gets placed.

To convert to a hard label, we apply a threshold:

$$\text{prediction} = \begin{cases} 1 & \text{if } \hat{y} \geq 0.5 \\ 0 & \text{if } \hat{y} < 0.5 \end{cases}$$

```
ŷ = 0.6832 ≥ 0.5  →  predict: 1 (placed)
true label: 0      →  WRONG — loss will be high
```

This error is then fed into **backpropagation**, which adjusts the weights to reduce the loss on the next pass.

---

## 8. Summary — Forward Pass at a Glance

```
Input X = [8.1, 92, 75, 76]
        ↓
Layer 1:   Z¹ = X · W¹ + b¹     →    A¹ = σ(Z¹)   ≈ [1.0, 1.0, 1.0]
        ↓
Layer 2:   Z² = A¹ · W² + b²    →    A² = σ(Z²)   ≈ [0.668, 0.668]
        ↓
Layer 3:   Z³ = A² · W³ + b³    →    A³ = σ(Z³)   ≈ 0.683
        ↓
Prediction: ŷ = 0.683  →  1 (placed)
```

Each layer does exactly the same thing — **linear transformation followed by sigmoid** — and the output of one layer feeds directly into the next.
