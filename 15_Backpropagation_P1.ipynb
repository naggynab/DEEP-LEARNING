# Backpropagation — Part 1

---

## 1. Introduction

**Backpropagation** (short for *backward propagation of errors*) is the algorithm used to train neural networks. It works by computing how much each weight and bias contributed to the total error, and then updating them to reduce that error.

It is always used together with an optimizer like **Gradient Descent**.

> **Core idea:** After making a prediction, we measure how wrong we were (loss), then work *backwards* through the network to figure out how much each parameter (weight/bias) is responsible for that error — and adjust them accordingly.

---

## 2. Network Architecture

We use a simple neural network with:
- **2 inputs:** `IQ` and `CGPA`
- **1 hidden layer** with 2 neurons
- **1 output neuron** predicting `LPA`

### Notation

| Symbol | Meaning |
|--------|---------|
| $x_1$ | Input: IQ |
| $x_2$ | Input: CGPA |
| $o_{11}$, $o_{12}$ | Outputs of hidden layer neurons 1 and 2 |
| $o_{21}$ | Final output (predicted LPA = $\hat{y}$) |
| $w^{1}_{11}$ | Weight from input $x_1$ to hidden neuron 1 |
| $w^{1}_{21}$ | Weight from input $x_2$ to hidden neuron 1 |
| $w^{1}_{12}$ | Weight from input $x_1$ to hidden neuron 2 |
| $w^{1}_{22}$ | Weight from input $x_2$ to hidden neuron 2 |
| $w^{2}_{11}$ | Weight from hidden neuron 1 to output |
| $w^{2}_{21}$ | Weight from hidden neuron 2 to output |
| $b_{11}$, $b_{12}$ | Biases for hidden neurons 1 and 2 |
| $b_{21}$ | Bias for output neuron |

### Diagram

![Diagram](./images/neural_network_backprop.svg)

```
         IQ (x₁)       CGPA (x₂)
            │                │
     w¹₁₁  │   w¹₂₁   w¹₁₂ │  w¹₂₂
         ┌──┴──────────────┐ │
         ▼                 ▼ ▼
    ┌─────────┐       ┌─────────┐
    │   o₁₁  │       │   o₁₂  │    ← Hidden Layer
    │ (+b₁₁) │       │ (+b₁₂) │
    └────┬────┘       └────┬────┘
         │    w²₁₁         │  w²₂₁
         └────────┬─────────┘
                  ▼
            ┌──────────┐
            │   o₂₁    │              ← Output Layer
            │  (+b₂₁)  │
            └──────────┘
                  │
                 ŷ (Predicted LPA)
```

### Forward Pass Equations

**Hidden neuron 1:**

$$o_{11} = w^{1}_{11} \cdot x_1 + w^{1}_{21} \cdot x_2 + b_{11}$$

**Hidden neuron 2:**

$$o_{12} = w^{1}_{12} \cdot x_1 + w^{1}_{22} \cdot x_2 + b_{12}$$

**Output neuron:**

$$\hat{y} = o_{21} = w^{2}_{11} \cdot o_{11} + w^{2}_{21} \cdot o_{12} + b_{21}$$

> Notice that $o_{21}$ depends on **5 things**: $w^{2}_{11}$, $w^{2}_{21}$, $o_{11}$, $o_{12}$, and $b_{21}$.
>
> And $o_{11}$ itself depends on **5 more things**: $w^{1}_{11}$, $w^{1}_{21}$, $x_1$ (IQ), $x_2$ (CGPA), and $b_{11}$.
>
> Since we are adjusting parameters starting from the *last* layer and going *backward*, this is called **backpropagation**.

## 3. Dataset

| IQ ($x_1$) | CGPA ($x_2$) | LPA ($y$) |
|------------|--------------|-----------|
| 80         | 8            | 3         |
| 60         | 9            | 5         |
| 70         | 5            | 8         |
| 120        | 7            | 11        |

## 4. Steps of Backpropagation

### Step 1 — Initialize Weights and Biases

We start by setting all weights to **1** and all biases to **0** (for simplicity):

$$w^{1}_{11} = w^{1}_{21} = w^{1}_{12} = w^{1}_{22} = w^{2}_{11} = w^{2}_{21} = 1$$

$$b_{11} = b_{12} = b_{21} = 0$$

---

### Step 2 — Select One Training Example

Let's pick the first student:

$$x_1 = 80 \quad (\text{IQ}), \qquad x_2 = 8 \quad (\text{CGPA}), \qquad y = 3 \quad (\text{actual LPA})$$

---

### Step 3 — Forward Propagation (Predict $\hat{y}$)

**Hidden neuron 1:**

$$o_{11} = w^{1}_{11} \cdot x_1 + w^{1}_{21} \cdot x_2 + b_{11}$$

$$o_{11} = 1 \times 80 + 1 \times 8 + 0 = 88$$

**Hidden neuron 2:**

$$o_{12} = w^{1}_{12} \cdot x_1 + w^{1}_{22} \cdot x_2 + b_{12}$$

$$o_{12} = 1 \times 80 + 1 \times 8 + 0 = 88$$

**Output neuron:**

$$\hat{y} = o_{21} = w^{2}_{11} \cdot o_{11} + w^{2}_{21} \cdot o_{12} + b_{21}$$

$$\hat{y} = 1 \times 88 + 1 \times 88 + 0 = 176$$

So the network predicts **LPA = 176**, but the actual value is **3**. That's a massive error — this is why we need to train the network!

---

### Step 4 — Calculate the Loss

We use **Mean Squared Error (MSE)** as our loss function:

$$L = (y - \hat{y})^2 = (3 - 176)^2 = (-173)^2 = 29{,}929$$

To reduce this loss, we need to update all 9 parameters:

$$w^{1}_{11},\ w^{1}_{21},\ w^{1}_{12},\ w^{1}_{22},\ w^{2}_{11},\ w^{2}_{21},\ b_{11},\ b_{12},\ b_{21}$$

---

### Step 5 — Update Weights via Gradient Descent

The update rules are:

$$w_{\text{new}} = w_{\text{old}} - \eta \cdot \frac{\partial L}{\partial w_{\text{old}}}$$

$$b_{\text{new}} = b_{\text{old}} - \eta \cdot \frac{\partial L}{\partial b_{\text{old}}}$$

Where $\eta$ (eta) is the **learning rate** — a small value like $0.01$ that controls how big each update step is.

To apply this, we need to find the derivative (gradient) of the loss $L$ with respect to every parameter.

---

## 5. Computing Gradients (Chain Rule)

We use the **chain rule** to break complex derivatives into smaller, easier pieces.

---

### 5.1 Derivative w.r.t. $w^{2}_{11}$ (output layer weight)

We want: $\dfrac{\partial L}{\partial w^{2}_{11}}$

By the chain rule:

$$\frac{\partial L}{\partial w^{2}_{11}} = \frac{\partial L}{\partial \hat{y}} \cdot \frac{\partial \hat{y}}{\partial w^{2}_{11}}$$

**Part 1** — derivative of loss w.r.t. prediction:

$$\frac{\partial L}{\partial \hat{y}} = \frac{\partial}{\partial \hat{y}}(y - \hat{y})^2 = -2(y - \hat{y})$$

Plugging in values: $y = 3$, $\hat{y} = 176$:

$$\frac{\partial L}{\partial \hat{y}} = -2(3 - 176) = -2 \times (-173) = 346$$

**Part 2** — derivative of prediction w.r.t. the weight:

$$\frac{\partial \hat{y}}{\partial w^{2}_{11}} = \frac{\partial}{\partial w^{2}_{11}}\left(w^{2}_{11} \cdot o_{11} + w^{2}_{21} \cdot o_{12} + b_{21}\right) = o_{11} = 88$$

**Combining:**

$$\boxed{\frac{\partial L}{\partial w^{2}_{11}} = 346 \times 88 = 30{,}448}$$

---

### 5.2 Derivative w.r.t. $w^{2}_{21}$ (output layer weight)

$$\frac{\partial L}{\partial w^{2}_{21}} = \frac{\partial L}{\partial \hat{y}} \cdot \frac{\partial \hat{y}}{\partial w^{2}_{21}} = -2(y - \hat{y}) \cdot o_{12}$$

$$= 346 \times 88 = \boxed{30{,}448}$$

---

### 5.3 Derivative w.r.t. $b_{21}$ (output bias)

$$\frac{\partial L}{\partial b_{21}} = \frac{\partial L}{\partial \hat{y}} \cdot \frac{\partial \hat{y}}{\partial b_{21}} = -2(y - \hat{y}) \cdot 1$$

$$= 346 \times 1 = \boxed{346}$$

---

### 5.4 Derivative w.r.t. $w^{1}_{11}$ (hidden layer weight — deeper layer)

This one goes one level deeper, so the chain is longer:

$$\frac{\partial L}{\partial w^{1}_{11}} = \frac{\partial L}{\partial \hat{y}} \cdot \frac{\partial \hat{y}}{\partial o_{11}} \cdot \frac{\partial o_{11}}{\partial w^{1}_{11}}$$

**Part 1** (already computed): $\dfrac{\partial L}{\partial \hat{y}} = 346$

**Part 2** — how does $\hat{y}$ change with $o_{11}$?

$$\frac{\partial \hat{y}}{\partial o_{11}} = \frac{\partial}{\partial o_{11}}\left(w^{2}_{11} \cdot o_{11} + w^{2}_{21} \cdot o_{12} + b_{21}\right) = w^{2}_{11} = 1$$

**Part 3** — how does $o_{11}$ change with $w^{1}_{11}$?

$$\frac{\partial o_{11}}{\partial w^{1}_{11}} = \frac{\partial}{\partial w^{1}_{11}}\left(w^{1}_{11} \cdot x_1 + w^{1}_{21} \cdot x_2 + b_{11}\right) = x_1 = 80$$

**Combining:**

$$\boxed{\frac{\partial L}{\partial w^{1}_{11}} = 346 \times 1 \times 80 = 27{,}680}$$

---

### 5.5 Summary of All 9 Gradients

Using the same chain rule approach for every parameter:

| Parameter | Gradient Formula | Value |
|-----------|-----------------|-------|
| $w^{2}_{11}$ | $-2(y-\hat{y}) \cdot o_{11}$ | $30{,}448$ |
| $w^{2}_{21}$ | $-2(y-\hat{y}) \cdot o_{12}$ | $30{,}448$ |
| $b_{21}$ | $-2(y-\hat{y})$ | $346$ |
| $w^{1}_{11}$ | $-2(y-\hat{y}) \cdot w^{2}_{11} \cdot x_1$ | $27{,}680$ |
| $w^{1}_{21}$ | $-2(y-\hat{y}) \cdot w^{2}_{11} \cdot x_2$ | $2{,}768$ |
| $w^{1}_{12}$ | $-2(y-\hat{y}) \cdot w^{2}_{21} \cdot x_1$ | $27{,}680$ |
| $w^{1}_{22}$ | $-2(y-\hat{y}) \cdot w^{2}_{21} \cdot x_2$ | $2{,}768$ |
| $b_{11}$ | $-2(y-\hat{y}) \cdot w^{2}_{11}$ | $346$ |
| $b_{12}$ | $-2(y-\hat{y}) \cdot w^{2}_{21}$ | $346$ |

---

### Step 5 Continued — Apply the Update (with $\eta = 0.0001$)

Using a small learning rate $\eta = 0.0001$ to avoid overshooting:

$$w^{2}_{11,\text{new}} = 1 - 0.0001 \times 30{,}448 = 1 - 3.04 = -2.04$$

$$w^{1}_{11,\text{new}} = 1 - 0.0001 \times 27{,}680 = 1 - 2.77 = -1.77$$

These large updates reflect the massive initial error. Over many iterations, the error shrinks and updates become smaller.

---

## 6. The Full Algorithm

```
Initialize all weights = 1, all biases = 0

For each training iteration:
    For each student (row) in the dataset:
        1. Forward pass:
               o₁₁ = w¹₁₁·x₁ + w¹₂₁·x₂ + b₁₁
               o₁₂ = w¹₁₂·x₁ + w¹₂₂·x₂ + b₁₂
               ŷ   = w²₁₁·o₁₁ + w²₂₁·o₁₂ + b₂₁

        2. Compute loss:
               L = (y - ŷ)²

        3. Backward pass (compute all 9 gradients using chain rule)

        4. Update all weights and biases:
               w_new = w_old - η · (∂L/∂w_old)
               b_new = b_old - η · (∂L/∂b_old)

Repeat until loss is sufficiently small
```

---

## 7. Key Takeaways

- **Backpropagation** computes gradients efficiently using the **chain rule**, going layer by layer from output back to input.
- **Each weight is updated** proportionally to how much it contributed to the error.
- **Learning rate** $\eta$ controls the step size — too large and it overshoots, too small and it converges slowly.
- In practice, this process runs over **thousands of iterations** and the loss gradually decreases toward zero.
- This note covers networks **without activation functions** (linear networks). In Part 2, we add non-linearities like ReLU or Sigmoid, which make the chain rule slightly more involved.
