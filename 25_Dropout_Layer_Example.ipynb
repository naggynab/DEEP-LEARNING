```python
X_train = np.linspace(-1, 1, 20)

y_train = np.array([-0.6561 , -0.3099 , -0.59035, -0.50855, -0.285  ,
                    -0.2443 , -0.02445,  0.00135, -0.2006 ,  0.07475,
                    -0.1422 ,  0.06515,  0.15265,  0.3521 ,  0.28415,
                    0.5524 ,  0.23115,  0.20835, 0.4211,  0.60485])

X_test = np.linspace(-1, 1, 20)

y_test = np.array([-0.69415, -0.451  , -0.43005, -0.4484 , -0.1475 ,
                   -0.5019 , -0.28055,  0.24595, -0.21425, -0.0286 ,
                   0.23415,  0.46575, 0.07955,  0.1973 ,  0.0719 ,
                   0.3639 ,  0.5536 ,  0.3365 , 0.50705,  0.33435])
```

# Regression Model Here

```python
model_1 = Sequential()
model_1.add(Dense(128, input_dim=1, activation="relu"))
model_1.add(Dense(128, activation="relu"))
model_1.add(Dense(1, activation="linear"))
adam = Adam(learning_rate=0.01)
model_1.compile(loss='mse', optimizer=adam, metrics=['mse'])
history = model_1.fit(X_train, y_train, epochs=500,
                    validation_data = (X_test, y_test),
                    verbose=False)
```

# Drawbacks of Dropout

> Dropout improves generalization, but comes with trade-offs worth understanding before applying it.

---

## 1. Slower Convergence

Because neurons are randomly disabled each epoch, the network receives noisier gradient updates and needs more epochs to converge compared to a network trained without dropout. Training time increases as a result.

## 2. Unstable Loss Curve

Since a different sub-network is active at every forward pass, the loss fluctuates between epochs rather than decreasing smoothly. This makes it harder to judge whether the model is truly converging or just oscillating.

## 3. Ineffective on Small Networks

Dropout reduces the effective capacity of the network during training. On small or shallow networks that are already limited in capacity, this can cause underfitting — the model becomes too restricted to learn the underlying patterns.

## 4. Interference with Batch Normalization

Applying dropout and Batch Normalization to the same layer can produce inconsistent variance estimates, leading to degraded performance. It is generally recommended to use one or the other, not both on the same layer.

## 5. Not Suitable for All Layer Types

Standard dropout does not work well with recurrent layers (LSTMs, GRUs) because dropping random timestep activations breaks the temporal dependencies the network is trying to learn. Specialized variants like **variational dropout** are used instead.

---

## Key Takeaway

> Dropout is a powerful tool, but it slows training and adds noise to the loss. Use it selectively — on large, overfit-prone networks — and avoid combining it with Batch Normalization on the same layer.
