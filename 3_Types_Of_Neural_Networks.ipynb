## 1. Types of Neural Networks 

Neural networks are not one-size-fits-all. Different architectures are designed for specific types of data and problems.

### 1.1 Artificial Neural Networks (ANN) / Multi-Layer Perceptron (MLP)
- Also called Feedforward Neural Networks.
- Consist of input layer, one or more hidden layers, and output layer.
- Fully connected layers.
- **Best suited for:** Tabular/structured data, regression, and classification tasks.
- Can learn complex non-linear relationships.
- Limitation: Not efficient for grid-like data (images) or sequential data.

### 1.2 Convolutional Neural Networks (CNN)
- Specialized for **grid-like data** such as images and videos.
- Use **Convolutional layers** + **Pooling layers** to automatically extract spatial features (edges → textures → objects).
- Key advantage: Parameter sharing and translation invariance.
- **Dominant in:** Computer Vision tasks (image classification, object detection, segmentation, facial recognition, medical imaging).

### 1.3 Recurrent Neural Networks (RNN)
- Designed for **sequential data** (time series, text, speech, video).
- Have loops/feedback connections that allow information to persist (memory).
- **Problems:** Vanishing/exploding gradients → difficulty learning long-term dependencies.
- **Improved variants:**
  - **LSTM** (Long Short-Term Memory)
  - **GRU** (Gated Recurrent Unit)
- **Applications:** Language modeling, machine translation, speech recognition, stock price prediction.

### 1.4 Autoencoders
- Unsupervised learning architecture.
- Consists of two parts:
  - **Encoder**: Compresses input into lower-dimensional latent representation.
  - **Decoder**: Reconstructs the input from the latent space.
- **Use cases:**
  - Dimensionality reduction
  - Denoising
  - Anomaly detection
  - Data compression
  - Generative tasks (Variational Autoencoders - VAE)

### 1.5 Generative Adversarial Networks (GANs)
- Framework introduced by Ian Goodfellow.
- Two networks competing:
  - **Generator**: Creates fake data.
  - **Discriminator**: Distinguishes real vs fake data.
- They improve through adversarial training (min-max game).
- **Famous for:** High-quality image generation, data augmentation, style transfer, deepfakes.

**Other notable architectures** (mentioned or implied): Transformers (later revolutionized NLP), Graph Neural Networks, etc.

## 2. History of Deep Learning 

### Early Years (1940s–1960s)
- **1943**: McCulloch-Pitts neuron model (theoretical).
- **1958**: Frank Rosenblatt’s **Perceptron** – First practical neural network. Created huge excitement.
- **1969**: Minsky & Papert’s book *Perceptrons* highlighted limitations (e.g., cannot solve XOR problem) → **First AI Winter**.

### Revival in 1980s
- **Backpropagation** algorithm popularized (Rumelhart, Hinton, Williams).
- Made training multi-layer networks feasible.
- Yann LeCun developed early **CNNs** (LeNet for handwritten digit recognition).

### 1990s–Early 2000s
- Development of **RNNs** and **LSTMs** (Hochreiter & Schmidhuber).
- Second slowdown due to limited computational power and data.

### Deep Learning Boom (2012 onwards)
- **2012**: **AlexNet** wins ImageNet competition by a huge margin → Deep Learning explosion.
- Key enablers:
  - GPUs for massive parallel computation
  - Large labeled datasets (ImageNet)
  - Better activation functions (ReLU)
  - Improved optimizers and regularization techniques
  - Open-source frameworks (TensorFlow, PyTorch)

Deep Learning shifted from academia to industry dominance.

## Key Takeaways

- **Choose the right architecture** based on data type:
  - Tabular → ANN/MLP
  - Images/Videos → CNN
  - Sequences → RNN/LSTM/GRU/Transformers
- Deep Learning success formula: **Large Data + Powerful Compute + Better Algorithms**.
- Understanding history helps appreciate current advancements.
- Theory provides perspective when implementing models in code.

## A More Concrete Mental Model

The main difference between architectures is **which relationships they assume are useful**:

| Architecture | Relationship it captures | Typical input | Example |
|---|---|---|---|
| ANN / MLP | Relationships between features in one example | A row of numbers | Predict customer churn |
| CNN | Nearby values and repeated local patterns | A grid of pixels | Classify a handwritten digit |
| RNN / LSTM / GRU | Order and context across time | A sequence | Predict the next word |
| Transformer | Relationships between every relevant part of a sequence | Tokens, patches, or multimodal inputs | Translate a paragraph |
| Autoencoder | A compact representation that can reconstruct the input | Images, vectors, or signals | Remove noise |
| GAN | A competition between generating and judging examples | Images or other data | Generate synthetic faces |

### Why one architecture is not automatically better

An MLP can technically receive the pixels of an image, but it treats every pixel as an unrelated input. It therefore has to learn locality and repeated visual patterns from scratch, which usually requires many more parameters and examples than a CNN.

Likewise, an RNN can process text one token at a time, but information must travel through many sequential steps. This makes long-range dependencies difficult to learn. LSTMs and GRUs add gates to control what information is kept, forgotten, and exposed. Transformers remove the sequential bottleneck by allowing tokens to attend directly to one another.

### Architecture-selection exercise

For each problem, identify the input structure first, then choose the architecture:

1. Predict house prices from area, rooms, age, and location.
2. Detect a tumor in an MRI scan.
3. Forecast electricity demand from hourly measurements.
4. Generate a summary of a long article.

A reasonable first pass is **MLP**, **CNN**, **LSTM/GRU or Transformer**, and **Transformer**, respectively. The data size, latency requirements, and need for interpretability can change the final choice, but the input structure is the first useful signal.
