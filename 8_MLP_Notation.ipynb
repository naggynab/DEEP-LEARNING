## 1. The Network Architecture

The network takes **4-dimensional input data** (n = 4 features per row).

```
Input layer (L0)     Hidden layer #1 (L1)     Hidden layer #2 (L2)     Output layer (L3)
   4 inputs               3 nodes                   2 nodes                 1 node

    x_i1  ──┐
    x_i2  ──┼──►  b_11  ──┐                                         
    x_i3  ──┤    b_12  ──┼──►  b_21  ──┐
    x_i4  ──┘    b_13  ──┘    b_22  ──┴──►  b_31  ──►  ŷ_i
```

![Actual Image](./images/8_1.png)

Data is represented as an **m × n matrix**:
- `m` = number of rows (training samples)
- `n` = 4 (number of features)
- Each row (one training sample) is denoted by `xᵢ`

The model parameters are collectively written as **(W, b, o)**.

## 3. Notation

### 3.1 Bias — `bᵢⱼ`

$$b_{ij}$$

- `i` = **layer** number
- `j` = **node number** within that layer

**Examples from the diagram:**

| Symbol | Meaning |
|---|---|
| `b₁₁` | 1st node of hidden layer 1 (L1) |
| `b₁₂` | 2nd node of hidden layer 1 (L1) |
| `b₁₃` | 3rd node of hidden layer 1 (L1) |
| `b₂₁` | 1st node of hidden layer 2 (L2) |
| `b₂₂` | 2nd node of hidden layer 2 (L2) |
| `b₃₁` | node of output layer (L3) |

---

### 3.2 Weights — `wᵏᵢⱼ`

$$w^k_{ij}$$

- `k` = **layer it is going to** (destination layer)
- `i` = **node it is coming from** (source node)
- `j` = **node it is going to** (destination node)

**Examples from the diagram:**

| Symbol | Meaning |
|---|---|
| `w¹₁₁` | Weight going to L1, from input 1, to node 1 |
| `w¹₄₂` | Weight going to L1, from input 4, to node 2 |
| `w¹₁₃` | Weight going to L1, from input 1, to node 3 |
| `w²` | All weights going to L2 (from L1 nodes to L2 nodes) |

The superscript `k` tells you which layer transition the weight belongs to — `w¹` weights are between L0→L1, `w²` between L1→L2, and so on.

---

### 3.3 Output — `oᵢⱼ`

$$o_{ij}$$

- `i` = **layer** number
- `j` = **node number** within that layer

**Examples from the diagram:**

| Symbol | Meaning |
|---|---|
| `o₁₁` | Output of node 1 in L1 |
| `o₁₂` | Output of node 2 in L1 |
| `o₁₃` | Output of node 3 in L1 |
| `o₂₁` | Output of node 1 in L2 |
| `o₂₂` | Output of node 2 in L2 |
| `o₃₁` | Output of the output layer = **ŷᵢ** (final prediction) |

The final output `o₃₁ = ŷᵢ` is the model's prediction for training sample `i`.
