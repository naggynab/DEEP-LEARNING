# MLP Intuition

---

## 1. The Core Problem

A single perceptron can only draw a **straight line** as its decision boundary. For non-linearly separable data (like XOR, or concentric circles), no single line will ever work.

The solution: **superimpose multiple perceptrons** to build a non-linear decision boundary out of several linear ones.

```
Perceptron 1:   /          Perceptron 2:   \          Combined:   /\
               /                            \                     /  \
   (one line)                  (another line)           (non-linear boundary)
```

---

## 2. How to Combine Two Perceptrons

Say you have two perceptrons, each producing a probability `p₁` and `p₂` via sigmoid. There are two ways to combine them:

### Option A — Equal combination (sum the probabilities)

$$\hat{y} = \sigma(p_1 + p_2)$$

Both perceptrons contribute equally. The result is passed through sigmoid again to squash it back to a probability.

### Option B — Weighted combination

$$\hat{y} = \sigma(w_1 p_1 + w_2 p_2 + b)$$

Each perceptron's output gets its own weight — the model learns how much to trust each perceptron. This is strictly more powerful than Option A.

**Option B is exactly what a hidden layer does.** The outputs of the first layer (`p₁`, `p₂`, ...) become inputs to the next layer, which combines them with learned weights.

---

## 3. Adding Hidden Layers

A **hidden layer** is just a set of perceptrons whose outputs are fed into the next layer rather than directly to the final output.

```
Input         Hidden Layer          Output
               (combines linear
               boundaries)

 x₁  ──┐                          
        ├──►  h₁  ──┐
 x₂  ──┤    h₂  ──┼──►  ŷ
        ├──►  h₃  ──┘
 x₃  ──┘
```

![Image for Hidden Layers](./images/9_1.png)

Each hidden node learns a different linear boundary. The output node then learns a **weighted combination** of those boundaries — producing a non-linear decision boundary overall.

**More hidden layers = more complex boundaries.** Each layer re-combines the outputs of the previous layer, allowing the network to build increasingly abstract representations:

| Layer | What it learns |
|---|---|
| Layer 1 | Simple edges, basic patterns |
| Layer 2 | Combinations of edges → shapes |
| Layer 3+ | Combinations of shapes → high-level concepts |

---

## 4. Why This Works — Geometric Intuition

Each perceptron draws a line and outputs a probability: values close to 1 on one side, close to 0 on the other.

When you **add** these outputs (with weights), you are essentially voting — each perceptron votes for which class a point belongs to. The final sigmoid turns that vote into a single probability.

The result is a boundary that can curve, bend, or form complex shapes depending on how many perceptrons contribute and what weights they are given.

---

## 5. Key Insight

> A single perceptron = one straight line.  
> Many perceptrons combined = any boundary shape.

This is why MLPs are called **universal function approximators** — given enough nodes and layers, they can approximate any continuous function to arbitrary precision. The perceptron trick alone can never do this.
